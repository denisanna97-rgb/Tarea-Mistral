## Celda 2: Importaciones

```
# pandas: Para trabajar con DataFrames (tablas de datos)
import pandas as pd

# files y userdata: Herramientas específicas de Google Colab
# - files.upload(): Sube archivos desde tu computadora
# - userdata.get(): Recupera API Keys de forma segura
from google.colab import files, userdata

# ChatMistralAI: Clase para conectarse a los modelos de Mistral AI
from langchain_mistralai.chat_models import ChatMistralAI

# IMPORTANTE: create_pandas_dataframe_agent se importa desde langchain_experimental
from langchain_experimental.agents import create_pandas_dataframe_agent

# warnings: Controla mensajes de advertencia
import warnings

# Ignoramos las advertencias para mantener la salida limpia
warnings.filterwarnings('ignore')

# Mensaje de confirmación
print("✅ Todas las bibliotecas se han importado correctamente")
print("📌 create_pandas_dataframe_agent importado desde langchain_experimental.agents")
```

### Celda 3: Subir el archivo CSV

```
# Mensaje para el usuario
print("📂 Por favor, selecciona tu archivo 'sales_data_sample.csv'")
print("   (Haz clic en 'Elegir archivos' y busca el archivo en tu computadora)")
print("")

# files.upload() abre un cuadro de diálogo para seleccionar archivos
# El resultado es un diccionario con {nombre_del_archivo: contenido}
uploaded = files.upload()

# Obtenemos el nombre del primer archivo subido
# list(uploaded.keys())[0] toma la primera clave del diccionario
nombre_archivo = list(uploaded.keys())[0]

# Confirmamos que el archivo se subió correctamente
print(f"\n✅ Archivo '{nombre_archivo}' subido correctamente")
```

### Celda 4: Cargar y Explorar los datos

```
# Cargamos el archivo CSV usando latin-1 (maneja caracteres especiales)
# latin-1 es una codificación que puede leer casi cualquier carácter
df = pd.read_csv(nombre_archivo, encoding='latin-1')

print("=" * 50)
print("📊 INFORMACIÓN DEL DATASET")
print("=" * 50)
print(f"📌 Número de filas: {df.shape[0]}")
print(f"📌 Número de columnas: {df.shape[1]}")
print(f"📌 Codificación usada: LATIN-1")
print("")

# === NOMBRES DE LAS COLUMNAS ===
print("📋 Nombres de las columnas:")
print(list(df.columns))
print("")

# === TIPOS DE DATOS ===
print("📝 Tipos de datos de cada columna:")
print(df.dtypes)
print("")

# === VISTA PREVIA ===
print("👀 Vista previa de los primeros 5 registros:")
print(df.head())
print("")

# === ESTADÍSTICAS BÁSICAS ===
print("📈 Estadísticas básicas (para columnas numéricas):")
print(df.describe())
print("")

print("✅ Dataset cargado correctamente")
```

### Celda 5: Configurar API Key de Mistral

```
print("=" * 50)
print("🔑 CONFIGURACIÓN DE API KEY")
print("=" * 50)
print("")
print("Instrucciones para crear el Secret en Colab:")
print("1. Haz clic en el ícono 🔑 'Secrets' en el panel lateral IZQUIERDO")
print("2. Haz clic en '+ Add new secret'")
print("3. En 'Name' escribe: MISTRAL_API_KEY")
print("4. En 'Value' pega tu API key de Mistral AI")
print("5. Activa el toggle (interruptor) para permitir el acceso")
print("")

# Recuperamos la API Key desde los Secrets de Colab
# userdata.get() es la forma SEGURA de acceder a las claves
mistral_api_key = userdata.get('MISTRAL_API_KEY')

# Verificamos si la API Key existe
if mistral_api_key:
    print("✅ API Key de Mistral encontrada correctamente")
    print(f"   (Los primeros 10 caracteres son: {mistral_api_key[:10]}...)")
else:
    print("❌ ERROR: No se encontró la API Key")
    print("   Asegúrate de haber creado el Secret 'MISTRAL_API_KEY'")
    print("   y haber activado el toggle de acceso")
```

### Celda 6: Inicializar y probar conexión con Mistral

```
print("=" * 50)
print("🤖 INICIALIZANDO MISTRAL AI")
print("=" * 50)
print("")

# Configuramos el modelo de Mistral
# model: "mistral-small-latest" es el modelo pequeño más reciente
# temperature: 0 = respuestas más precisas (menos creativas)
# api_key: Nuestra clave de API
llm = ChatMistralAI(
    model="mistral-small-latest",    # Modelo de Mistral a usar
    temperature=0,                    # Precisión sobre creatividad
    api_key=mistral_api_key           # Nuestra API Key
)

print("✅ Modelo 'mistral-small-latest' configurado")
print("")

# === PROBAR CONEXIÓN ===
print("🔄 Probando conexión con Mistral AI...")

try:
    # Enviamos un mensaje de prueba al modelo
    # invoke() envía el mensaje y espera la respuesta
    respuesta = llm.invoke("Responde exactamente: ¡Conexión exitosa! 🚀")

    # Si llegamos aquí, la conexión funcionó
    print(f"✅ {respuesta.content}")
    print("")
    print("🎉 ¡La conexión con Mistral AI es exitosa!")

except Exception as error:
    # Si hay algún error, lo mostramos
    print(f"❌ Error de conexión: {error}")
    print("   Verifica tu API Key y tu conexión a internet")
```

### Celda 7: Crear el Agente de Pandas

```
print("=" * 50)
print("⚙️ CREANDO EL AGENTE DE PANDAS")
print("=" * 50)
print("")

# Creamos el agente que interactuará con nuestro DataFrame
# Este agente entiende lenguaje natural y lo convierte en código Python/pandas

agente = create_pandas_dataframe_agent(
    llm=llm,                         # El modelo de Mistral que configuramos
    df=df,                           # Nuestro DataFrame con los datos de ventas
    verbose=True,                    # Muestra el razonamiento del agente
    allow_dangerous_code=True,       # Permite que el agente ejecute código pandas
    handle_parsing_errors=True,      # Maneja errores de formato automáticamente
)

print("✅ Agente creado exitosamente")
print("")
print("📌 El agente puede:")
print("   - Responder preguntas sobre los datos")
print("   - Realizar cálculos (sumas, promedios, máximos, etc.)")
print("   - Filtrar y agrupar información")
print("   - Generar análisis en lenguaje natural")
```

### Celda 8: Probar el Agente

```
print("=" * 50)
print("🚀 PROBANDO EL AGENTE")
print("=" * 50)
print("")

# Función para hacer preguntas al agente
def preguntar(pregunta):
    """Envía una pregunta al agente y muestra la respuesta"""
    print(f"\n👤 TÚ: {pregunta}")
    print("🤖 AGENTE: ", end="", flush=True)  # flush=True muestra inmediatamente

    # El agente procesa la pregunta y genera una respuesta
    respuesta = agente.run(pregunta)

    print(respuesta)
    print("-" * 50)
    return respuesta

# === PRIMERAS PREGUNTAS ===

print("Realizando pruebas...")
print("")

# Pregunta 1: Información básica
preguntar("¿Cuántas filas y cuántas columnas tiene este dataset?")

# Pregunta 2: Estructura de los datos
preguntar("¿Cuáles son los nombres de todas las columnas?")

# Pregunta 3: Análisis simple
preguntar("¿Cuál es el total de ventas (SALES) sumando todos los registros?")

# Pregunta 4: Análisis por categoría
preguntar("¿Cuáles son los 3 productos (PRODUCTLINE) más vendidos?")

print("\n" + "=" * 50)
print("🎉 ¡Todas las pruebas completadas exitosamente!")
print("=" * 50)
print("")
print("📝 El agente está listo para responder cualquier pregunta")
print("   sobre tus datos de ventas en lenguaje natural.")
```

### Celda 9: Interacción continua (opcional)

```
print("=" * 50)
print("💬 MODO INTERACTIVO")
print("=" * 50)
print("")
print("Escribe tus preguntas sobre los datos de ventas.")
print("Escribe 'salir' o 'exit' para terminar.")
print("")

# Bucle infinito para seguir preguntando
while True:
    # Pedimos al usuario que escriba una pregunta
    pregunta_usuario = input("\n👤 Tu pregunta: ")

    # Verificamos si quiere salir
    if pregunta_usuario.lower() in ['salir', 'exit', 'q', 'quit']:
        print("\n👋 ¡Hasta luego!")
        break

    # Si la pregunta no está vacía
    if pregunta_usuario.strip():
        print("🤖 AGENTE: ", end="", flush=True)
        try:
            respuesta = agente.run(pregunta_usuario)
            print(respuesta)
        except Exception as e:
            print(f"\n❌ Error al procesar la pregunta: {e}")
            print("   Intenta reformular tu pregunta.")
    else:
        print("⚠️ Por favor, escribe una pregunta válida.")
```
